```python
#Indexing
arr=numpy.random.randint(1,10000, 5)
print(arr)
print("arr[::-1] => ",arr[::-1])
```

```python
#Indexing 2 D array

arr=numpy.random.randint(1,10000, size=(4,3))
print(arr)
print("arr[[0]]=>", arr[[0]])  # 0rd row
print("arr[[0]]=>", arr[[0]])   # 3rd row
print("arr[[0,3]]=>", arr[[0,3]])   # 0 and 3rd row
print("arr[0][2] =>", arr[0][2])   # 0 row and 2 index in 0 row

print("arr[:,[1]] =>", arr[:,[1]])   # 1 st column
print("arr[:,[1]] =>", arr[:,[1,2]])   # 1 and 3 rd column

```
